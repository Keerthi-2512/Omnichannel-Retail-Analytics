```python
#INVENTORY RISK ANALYSIS
#CREATE STOCK STATUS
df_final['Stock_Status'] = df_final.apply(
    lambda x: 'Low Stock' if x['inventory_level'] < x['reorder_point'] else 'Safe',
    axis=1
)
```

```python
#COUNT RISK PRODUCTS
df_final['Stock_Status'].value_counts().plot(kind='bar')
plt.title('Stock Risk Distribution')
plt.show()
```

```python
The stock risk distribution shows that the majority of products are in a safe inventory state, indicating overall stability in 
inventory management.However, a significant portion of products fall under the low stock category, highlighting potential risks 
of stockouts and lost sales.This imbalance also affects machine learning performance, as models tend to favor the majority class 
(safe products), making it important to handle class imbalance properly.
```

```python
#COUNT RISK PRODUCTS
risk_percent = (df_final['Stock_Status']=='Low Stock').mean() * 100
print(f"Risk Percentage: {risk_percent:.2f}%")
```

```python
#SALES vs INVENTORY
plt.figure(figsize=(8,5))
plt.scatter(df_final['inventory_level'], df_final['Sales'])
plt.xlabel('Inventory Level')
plt.ylabel('Sales')
plt.title('Sales vs Inventory Analysis')
plt.show()
```

```python
The scatter plot shows no strong relationship between inventory levels and sales, indicating that stock availability is not aligned with demand.
Many products have high inventory but low sales, suggesting overstocking and inefficient inventory planning. Additionally, only a few products 
generate high sales, indicating that demand is concentrated among specific items.This highlights the need for demand-driven inventory 
management to optimize stock levels and reduce holding costs.
```

```python
#QUADRANT ANALYSIS
sales_avg = df_final['Sales'].mean()
inventory_avg = df_final['inventory_level'].mean()

def category(row):
    if row['Sales'] >= sales_avg and row['inventory_level'] >= inventory_avg:
        return 'High Demand & High Stock'
    elif row['Sales'] >= sales_avg and row['inventory_level'] < inventory_avg:
        return 'High Demand & Low Stock'
    elif row['Sales'] < sales_avg and row['inventory_level'] >= inventory_avg:
        return 'Low Demand & High Stock'
    else:
        return 'Low Demand & Low Stock'

df_final['Category'] = df_final.apply(category, axis=1)
```

```python
#VISUALIZE QUADRANTS
import seaborn as sns

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df_final,
    x='inventory_level',
    y='Sales',
    hue='Category'
)

plt.axhline(sales_avg, color='black')
plt.axvline(inventory_avg, color='black')

plt.title('Sales vs Inventory Quadrant')
plt.show()
```

```python
The quadrant analysis reveals that inventory is not optimally aligned with demand.Several products fall under high demand but low stock, 
indicating potential lost sales opportunities due to insufficient inventory.At the same time, a significant number of products are overstocked
despite low demand, leading to inefficient inventory utilization and increased holding costs.
This highlights the need for demand-driven inventory planning and better stock allocation strategies.
```

```python
Accuracy: 0.85  (looks good)
BUT Low Stock recall = 0.01  (VERY BAD). because data is imbalance.
Safe → 90,627
Low Stock → 16,348
```

```python
USE CLASS WEIGHT
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced',
    random_state=42
)
```

```python
This forces model to care about minority class
```

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
After handling class imbalance using class weights, the model improved significantly in identifying low stock situations. 
While overall accuracy decreased, recall for critical low stock cases increased to 60%, making the model more useful for business decision-making.

This highlights the importance of focusing on recall over accuracy in imbalanced classification problems.
```
